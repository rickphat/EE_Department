```python
# hangman.py
# Name:
# Collaborators:
# Time spent:

import random
import string

WORDLIST_FILENAME = "words.txt"


def load_words():
    """
    Returns a list of valid words.
    """
    print("Loading word list from file...")
    inFile = open(WORDLIST_FILENAME, 'r')
    line = inFile.readline()
    wordlist = line.split()
    print("  ", len(wordlist), "words loaded.")
    return wordlist


def choose_word(wordlist):
    """
    wordlist (list): list of words
    Returns a word from wordlist at random
    """
    return random.choice(wordlist)


# Load word list
wordlist = load_words()


def is_word_guessed(secret_word, letters_guessed):
    '''
    Returns True if all letters in secret_word
    have been guessed.
    '''
    for letter in secret_word:
        if letter not in letters_guessed:
            return False

    return True


def get_guessed_word(secret_word, letters_guessed):
    '''
    Returns the secret word with unguessed letters
    represented by "_ ".
    '''
    result = ""

    for letter in secret_word:
        if letter in letters_guessed:
            result += letter + " "
        else:
            result += "_ "

    return result


def get_available_letters(letters_guessed):
    '''
    Returns all lowercase letters that have not
    been guessed yet.
    '''
    result = ""

    for letter in string.ascii_lowercase:
        if letter not in letters_guessed:
            result += letter

    return result


def hangman(secret_word):
    '''
    Starts the Hangman game.
    '''

    guesses_remaining = 6
    warnings_remaining = 3
    letters_guessed = []

    print("Welcome to the game Hangman!")
    print("I am thinking of a word that is",
          len(secret_word), "letters long.")
    print("You have", warnings_remaining, "warnings left.")
    print("You have", guesses_remaining, "guesses left.")

    while guesses_remaining > 0:

        print("-------------")
        print("You have", guesses_remaining, "guesses left.")
        print("Available letters:",
              get_available_letters(letters_guessed))

        guess = input("Please guess a letter: ").lower()

        # Invalid input
        if len(guess) != 1 or not guess.isalpha():

            if warnings_remaining > 0:
                warnings_remaining -= 1

                if warnings_remaining == 1:
                    print("Oops! That is not a valid letter. "
                          "You have 1 warning left:",
                          get_guessed_word(secret_word, letters_guessed))
                else:
                    print("Oops! That is not a valid letter. "
                          "You have", warnings_remaining,
                          "warnings left:",
                          get_guessed_word(secret_word, letters_guessed))
            else:
                guesses_remaining -= 1
                print("Oops! That is not a valid letter. "
                      "You have no warnings left "
                      "so you lose one guess:",
                      get_guessed_word(secret_word, letters_guessed))

            continue

        # Already guessed
        if guess in letters_guessed:

            if warnings_remaining > 0:
                warnings_remaining -= 1

                print("Oops! You've already guessed that letter.",
                      "You have", warnings_remaining,
                      "warnings left:",
                      get_guessed_word(secret_word, letters_guessed))
            else:
                guesses_remaining -= 1

                print("Oops! You've already guessed that letter.",
                      "You have no warnings left "
                      "so you lose one guess:",
                      get_guessed_word(secret_word, letters_guessed))

            continue

        # Add new guess
        letters_guessed.append(guess)

        # Correct guess
        if guess in secret_word:
            print("Good guess:",
                  get_guessed_word(secret_word, letters_guessed))

        # Incorrect guess
        else:
            print("Oops! That letter is not in my word:",
                  get_guessed_word(secret_word, letters_guessed))

            # Vowels cost 2 guesses
            if guess in "aeiou":
                guesses_remaining -= 2
            else:
                guesses_remaining -= 1

        # Check if player won
        if is_word_guessed(secret_word, letters_guessed):
            print("Congratulations, you won!")

            unique_letters = len(set(secret_word))
            score = guesses_remaining * unique_letters

            print("Your total score for this game is:", score)
            return

    # Player lost
    print("Sorry, you ran out of guesses.")
    print("The word was", secret_word + ".")


def match_with_gaps(my_word, other_word):
    '''
    Returns True if other_word matches my_word.
    '''

    # Remove spaces
    my_word = my_word.replace(" ", "")

    # Different lengths cannot match
    if len(my_word) != len(other_word):
        return False

    # Find letters that have already been revealed
    revealed_letters = []

    for letter in my_word:
        if letter != "_":
            revealed_letters.append(letter)

    # Check each position
    for i in range(len(my_word)):

        # If there is a revealed letter,
        # it must match exactly.
        if my_word[i] != "_":

            if my_word[i] != other_word[i]:
                return False

        # If there is a blank, it cannot contain
        # a letter that has already been revealed.
        else:

            if other_word[i] in revealed_letters:
                return False

    return True


def show_possible_matches(my_word):
    '''
    Prints all words in wordlist that match my_word.
    '''

    matches = []

    for word in wordlist:

        if match_with_gaps(my_word, word):
            matches.append(word)

    if len(matches) == 0:
        print("No matches found.")
    else:
        print("Possible word matches are:")

        for word in matches:
            print(word, end=" ")

        print()


def hangman_with_hints(secret_word):
    '''
    Hangman game with * hint functionality.
    '''

    guesses_remaining = 6
    warnings_remaining = 3
    letters_guessed = []

    print("Welcome to the game Hangman!")
    print("I am thinking of a word that is",
          len(secret_word), "letters long.")
    print("You have", guesses_remaining, "guesses left.")

    while guesses_remaining > 0:

        print("-------------")
        print("You have", guesses_remaining, "guesses left.")
        print("Available letters:",
              get_available_letters(letters_guessed))

        guess = input("Please guess a letter: ").lower()

        # Hint
        if guess == "*":

            current_word = get_guessed_word(
                secret_word,
                letters_guessed
            )

            show_possible_matches(current_word)

            continue

        # Invalid input
        if len(guess) != 1 or not guess.isalpha():

            if warnings_remaining > 0:
                warnings_remaining -= 1

                print("Oops! That is not a valid letter.",
                      "You have", warnings_remaining,
                      "warnings left:",
                      get_guessed_word(secret_word, letters_guessed))

            else:
                guesses_remaining -= 1

                print("Oops! That is not a valid letter.",
                      "You have no warnings left "
                      "so you lose one guess:",
                      get_guessed_word(secret_word, letters_guessed))

            continue

        # Already guessed
        if guess in letters_guessed:

            if warnings_remaining > 0:
                warnings_remaining -= 1

                print("Oops! You've already guessed that letter.",
                      "You have", warnings_remaining,
                      "warnings left:",
                      get_guessed_word(secret_word, letters_guessed))

            else:
                guesses_remaining -= 1

                print("Oops! You've already guessed that letter.",
                      "You have no warnings left "
                      "so you lose one guess:",
                      get_guessed_word(secret_word, letters_guessed))

            continue

        # Add new letter
        letters_guessed.append(guess)

        # Correct guess
        if guess in secret_word:

            print("Good guess:",
                  get_guessed_word(secret_word, letters_guessed))

        # Incorrect guess
        else:

            print("Oops! That letter is not in my word:",
                  get_guessed_word(secret_word, letters_guessed))

            # Vowels cost 2 guesses
            if guess in "aeiou":
                guesses_remaining -= 2
            else:
                guesses_remaining -= 1

        # Check win
        if is_word_guessed(secret_word, letters_guessed):

            print("Congratulations, you won!")

            unique_letters = len(set(secret_word))
            score = guesses_remaining * unique_letters

            print("Your total score for this game is:", score)

            return

    # Lose
    print("Sorry, you ran out of guesses.")
    print("The word was", secret_word + ".")


if __name__ == "__main__":

    # Normal Hangman
    secret_word = choose_word(wordlist)
    hangman(secret_word)

    # For Hangman with hints, comment out the two lines above
    # and uncomment these:
    #
    # secret_word = choose_word(wordlist)
    # hangman_with_hints(secret_word)
```
