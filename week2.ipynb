````python
title = "Space Explorer"
launches = 3
fuel_percentage = 87.5
is_ready = True
crew = ["ari", "bo", "bb8"]
ship = {"name":"SpaceExplorer", "speed":"9000"}

for values in [title, launches, fuel_percentage, is_ready, crew, ship]:
  print(repr(values) , "->", type(values).__name__)
````
